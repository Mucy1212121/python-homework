### ข้อ 3. จงเขียนโปรแกรมคำนวณพื้นที่สี่เหลี่ยมผืนผ้า โดยรับค่าความกว้างและความยาวจากผู้ใช้ (2 คะแนน)

*ตัวอย่างการทำงาน:*
```
ป้อนความกว้าง: 5
ป้อนความยาว: 8

ผลลัพธ์:
พื้นที่สี่เหลี่ยมผืนผ้า = 40.00 ตารางหน่วย
```

````python
# พิมพ์โค้ดคำตอบที่นี่
a = int(input("ป้อนความกว้าง: "))
b = int(input("ป้อนความยาว: "))

c = a * b 
print("\nผลลัพธ์:")
print(f"พื้นที่สี่เหลี่ยมผืนผ้า = {c:.2f} ตารางหน่วย")
````

### ข้อ 4. กำหนดให้ List ของคะแนนสอบนักศึกษา ดังนี้ (2 คะแนน)

```python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
```

จงเขียนโปรแกรมเพื่อ:
- a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)

````python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
# a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)
a = sum(scores) / len(scores)
print(f"แสดงคะแนนเฉลี่ย = {a:.2f}")
````

### ข้อ 5. กำหนดให้ List ของชื่อนักศึกษา ดังนี้ (2 คะแนน)

```python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]
```

จงเขียนโปรแกรมเพื่อ:
- a) เพิ่มชื่อ "สมพร" ต่อท้าย List
- b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
- c) ลบชื่อ "สมศรี" ออกจาก List
- d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด

````python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]

# a) เพิ่มชื่อ "สมพร" ต่อท้าย List
students.append('สมพร')

# b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
students.insert(2, "สมใจ")

# c) ลบชื่อ "สมศรี" ออกจาก List
students.remove("สมศรี")

# d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด
print(students)
````
